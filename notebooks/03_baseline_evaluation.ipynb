# EduTune AI — Baseline Evaluation

Inspect baseline model readiness and stored evaluation metrics.

**Important:** the current CPU-only environment blocked Mistral-7B loading; stored perfect metric values describe the stored prediction/reference artifact, not successful Mistral-7B inference.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

def load_jsonl(path):
    return pd.read_json(path, lines=True)

print("Project root:", PROJECT_ROOT)

E = PROJECT_ROOT / "data/evaluation"
with open(E / "baseline_model_report.json", encoding="utf-8") as f:
    model_report = json.load(f)
with open(E / "baseline_evaluation_report.json", encoding="utf-8") as f:
    evaluation = json.load(f)
display(pd.DataFrame([model_report]))
display(pd.DataFrame([evaluation.get("metrics", {})]))
```

## Item-Level Results

```python
results = pd.DataFrame(evaluation.get("results", []))
display(results)
```
