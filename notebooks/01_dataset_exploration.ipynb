# EduTune AI — Dataset Exploration

Explore the raw, curated, and synthetic educational datasets used by EduTune AI.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

def load_jsonl(path):
    return pd.read_json(path, lines=True)

print("Project root:", PROJECT_ROOT)

raw = load_jsonl(PROJECT_ROOT / "data/raw/education_seed.jsonl")
curated = load_jsonl(PROJECT_ROOT / "data/processed/curated_dataset.jsonl")
synthetic = load_jsonl(PROJECT_ROOT / "data/synthetic/synthetic_dataset.jsonl")
print("Raw:", len(raw), "Curated:", len(curated), "Synthetic:", len(synthetic))
```

## Dataset Schema and Samples

```python
for name, df in [("Raw", raw), ("Curated", curated), ("Synthetic", synthetic)]:
    print(f"\n{name}: shape={df.shape}")
    print("Columns:", list(df.columns))
    display(df.head())
```

## Category, Difficulty, and Task-Type Distributions

```python
for name, df in [("Raw", raw), ("Curated", curated), ("Synthetic", synthetic)]:
    print(f"\n{name}")
    for col in ["category", "difficulty", "task_type"]:
        if col in df.columns:
            display(df[col].value_counts().rename("count").to_frame())
```

```python
for col in ["category", "difficulty", "task_type"]:
    if col in synthetic.columns:
        synthetic[col].value_counts().plot(kind="bar")
        plt.title(f"Synthetic Dataset — {col.replace('_', ' ').title()}")
        plt.xlabel(col.replace("_", " ").title())
        plt.ylabel("Records")
        plt.xticks(rotation=30, ha="right")
        plt.tight_layout()
        plt.show()
```

## Basic Quality Checks

```python
required = {"instruction", "response", "category"}
for name, df in [("Raw", raw), ("Curated", curated), ("Synthetic", synthetic)]:
    print(f"{name}: missing={df.isna().sum().sum()}, duplicate_rows={df.duplicated().sum()}")
    print("Missing required columns:", sorted(required - set(df.columns)))
    for col in required & set(df.columns):
        print(f"Empty {col}:", df[col].fillna("").astype(str).str.strip().eq("").sum())
```
